# RSA Playground
Generate a small RSA key pair and demonstrate encryption, decryption,
and signing.

```python
from cryptographic_algorithms.asymmetric import rsa, digital_signature
key_pair = rsa.generate_keypair(256)
message = b'cryptography'
ciphertext = rsa.encrypt_bytes(message, key_pair.public_key)
recovered = rsa.decrypt_bytes(ciphertext, key_pair)
recovered
```

```python
scheme = digital_signature.RSASignatureScheme(key_pair)
signature = scheme.sign(message)
scheme.verify(message, signature)
```
