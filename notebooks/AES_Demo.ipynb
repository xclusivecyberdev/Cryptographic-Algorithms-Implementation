# AES Interactive Demonstration
This notebook walks through a single AES encryption round using the 
educational implementation contained in this repository.

```python
from cryptographic_algorithms.symmetric.aes import AES
key = bytes.fromhex('000102030405060708090a0b0c0d0e0f')
cipher = AES(key)
plaintext = bytes.fromhex('00112233445566778899aabbccddeeff')
ciphertext = cipher.encrypt_block(plaintext)
ciphertext.hex()
```

```python
cipher.decrypt_block(ciphertext).hex()
```
